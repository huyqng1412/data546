```python
# Many Machine Learning problems involve thousands or 
# even millions of features for each training instance. 
# Not only do all these features make training extremely slow,
# but they can also make it much harder to find a good solution, as we will see. 
# This problem is often referred to as the curse of dimensionality.
```

```python
# Reducing dimensionality does cause some information loss (just
# like compressing an image to JPEG can degrade its quality), 
# so even though it will speed up training, it may make your system 
# perform slightly worse. It also makes your pipelines a bit more
# complex and thus harder to maintain. So, if training is too slow,
# you should first try to train your system with the original data
# before considering using dimensionality reduction.
```

```python
# In some cases, reducing the dimensionality of the training data may filter out
# some noise and unnecessary details and thus result in higher performance,
# but in general it won’t; it will just speed up training.
```

```python
# The core idea of the term "Curse of Dimensionality" in ML is that when 
# we work with data in high dimensional spaces, 
# if the data has so many features, and its dimensions are too big, 
# it could make the data too sparse, and it becomes much difficult to analyze, or
# visualize. High dimensional data can leads to overfitting because it has too many noises. 
```

```python
# Manifold learning
```

```python
# A type of non-linear dimensionality reduction.
# It is an assumption that most real-world high-dimensional datasets 
# lie close to a much lower-dimensional manifold.
# The Swiss roll is split into two classes: in the 3D space (on the left), 
# the decision boundary would be fairly complex, 
# but in the 2D unrolled manifold space (on the right), 
# the decision boundary is a straight line.

# But it not always true, as shown in the lower figure of 8-6. 
# This decision boundary looks very simple in the original 3D space 
# (a vertical plane), but it looks more complex
# in the unrolled manifold (a collection of four independent line segments). 
```

```python
# PCA
# Principal Component Analysis (PCA) is by far the most 
# popular dimensionality reduction algorithm. 
# First it identifies the hyperplane that lies closest to the data, and then
# it projects the data onto it, just like in Figure 8-2.
```

```python
# Preserving the Variance
```

```python
# Before you can project the training set onto a lower-dimensional hyperplane, 
# you first need to choose the right hyperplane.

# A preserving line that contains the most data instances, as shown in 8-7, 
# the solid line preserves the maximum variance, while the other have less.

# It seems reasonable to select the axis that preserves the maximum amount of 
# variance, as it will most likely lose less information than the other projections.
```

```python
# Principal Components
```

```python
# Principle Components Axis identifies the axis that accounts for the largest amount of 
# variance in the training set.

# It also finds a second axis, orthogonal to the first one, 
# that accounts for the largest amount of remaining variance. 
# In this 2D example there is no choice: it is the dotted line.
```

```python
# The ith axis is called the ith principal component (PC) of the data.
# In Figure 8-7, the first PC is the axis on which vector c1 lies, 
# and the second PC is the orthogonal axis on which vector c2 lies.

# The first principal component (PC1) is the direction of maximum variance 
# (the line that best explains the spread of the data).

# The second principal component (PC2) is the direction of the next 
# highest variance, orthogonal to the first.

# There is a standard matrix factorization technique called 
# Singular Value Decomposition (SVD) that can decompose the 
# training set matrix X into the matrix multiplication of three
# matrices U Σ V⊺, where V contains the unit vectors that 
# define all the principal components that we are looking for, 
# as shown in Equation 8-1.
```

```python
# Once you have identified all the principal components, 
# you can reduce the dimensionality of the dataset down to d dimensions 
# by projecting it onto the hyperplane 
# defined by the first d principal components.

# Projection = shadow of high-dimensional data onto fewer dimensions.
# Keep the top d principal components that explain the most variance.
# 2D of a 3D data variances against a hyperplane in a direction that we 
# can mine out the most information but does not change the structure.

# To project the training set onto the hyperplane and 
# obtain a reduced dataset Xd-proj of dimensionality d, 
# compute the matrix multiplication of the training set matrix X by
# the matrix Wd, defined as the matrix containing the first d columns of V. 
```
